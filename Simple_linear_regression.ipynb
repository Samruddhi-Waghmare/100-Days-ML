```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('placement.csv',usecols=['cgpa','iq'])
```

```python
df.head()
```

```python
plt.scatter(df['cgpa'],df['iq'])
plt.xlabel('CGPA')
plt.ylabel('IQ')
```

```python
X = df.iloc[:,0:1]
y = df.iloc[:,-1]
```

```python
X
#input column
```

```python
y 
#output column
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train,y_train)
```

```python
X_test
```

```python
y_test
```

```python
lr.predict(X_test.iloc[3].values.reshape(1,1))
```

```python
plt.scatter(df['cgpa'],df['iq'])
plt.plot(X_test,lr.predict(X_test),colour='red')
plt.xlabel('CGPA')
plt.ylabel('IQ')
```
